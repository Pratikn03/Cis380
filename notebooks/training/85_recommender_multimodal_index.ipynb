## Visuals and metrics


```python
import matplotlib.pyplot as plt
from app.models.recommender.predict import predict_multimodal

try:
    result = predict_multimodal(text='red sneakers', image_bytes=None, top_k=5)
    items = result.get('items', [])
    if not items:
        print('No items returned')
    else:
        titles = [i.get('title', 'item') for i in items]
        scores = [i.get('score', 0.0) for i in items]
        plt.figure(figsize=(6, 3))
        plt.barh(titles, scores)
        plt.title('Top multimodal matches')
        plt.xlabel('score')
        plt.tight_layout()
        plt.show()
except Exception as exc:
    print('Multimodal search failed:', exc)
```
